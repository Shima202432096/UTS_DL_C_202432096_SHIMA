  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>

```python
#202432096_SHIMA
import os
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

# Seed agar hasil dapat direproduksi
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device yang digunakan:', device)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432096_SHIMA
# 1. Import Dtaset

DATA_PATH = 'kuala-lumpur-air-quality.csv'
RAW_URL = 'https://raw.githubusercontent.com/Shima202432096/UTS_DL_C_202432096_SHIMA/main/kuala-lumpur-air-quality.csv'
df = pd.read_csv(DATA_PATH if os.path.exists(DATA_PATH) else RAW_URL, skipinitialspace=True)
print('Jumlah baris, kolom :', df.shape)

# 2. Ubah tipe data: tanggal -> datetime, AQI -> numerik (sel kosong menjadi NaN)
df['date'] = pd.to_datetime(df['date'], format='%Y/%m/%d')
df['aqi'] = pd.to_numeric(df['aqi'], errors='coerce')

# 3. Urutkan kronologis dan jadikan tanggal sebagai index
df = df.sort_values('date').set_index('date')
print('Rentang tanggal     :', df.index.min().date(), 's.d.', df.index.max().date())
print('Nilai AQI kosong    :', df['aqi'].isna().sum())
print(df['aqi'].describe())

# 4. Visualisasi data mentah
plt.figure(figsize=(12, 4))
plt.plot(df.index, df['aqi'], color='#114F6A', linewidth=0.8)
plt.title('AQI Harian Kuala Lumpur (Data Mentah)')
plt.xlabel('Tanggal')
plt.ylabel('AQI')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
df.head()
```

```python
#202432096_SHIMA
# Pembersihan data: tangani hari yang hilang dan nilai kosong
MAX_GAP = 7     # celah <= 7 hari diisi interpolasi linear, celah lebih panjang dibiarkan kosong
STUCK_MIN = 10  # nilai identik >= 10 hari berturut-turut dianggap artefak pencatatan (bukan kondisi udara nyata)

# 1. Buat kalender harian lengkap agar tanggal yang hilang terlihat sebagai NaN
full = df.reindex(pd.date_range(df.index.min(), df.index.max(), freq='D'))

# 1b. Deteksi dan hapus nilai konstan yang terlalu panjang (misalnya AQI = 52 selama 15 Des 2019 - 17 Jan 2020)
v = full['aqi']
grp = (v != v.shift()).cumsum()
stuck = v.notna() & (v.groupby(grp).transform('size') >= STUCK_MIN)
print('Hari AQI konstan (artefak) dihapus:', stuck.sum())
full.loc[stuck, 'aqi'] = np.nan

is_na = full['aqi'].isna()
print('Total hari kalender :', len(full))
print('Hari tanpa nilai    :', is_na.sum())

# 2. Hitung panjang setiap celah (deretan NaN berurutan)
run_id = (is_na != is_na.shift()).cumsum()
run_len = is_na.groupby(run_id).transform('sum')
short_gap = is_na & (run_len <= MAX_GAP)

# 3. Isi celah pendek dengan interpolasi linear; celah panjang tetap NaN
interp = full['aqi'].interpolate(method='linear', limit_area='inside')
full['aqi_clean'] = full['aqi'].where(~short_gap, interp)
print('Diisi interpolasi   :', short_gap.sum())
print('Tetap kosong        :', full['aqi_clean'].isna().sum(), '(celah panjang, tidak dipakai untuk membentuk window)')

# 4. Visualisasi setelah pembersihan
plt.figure(figsize=(12, 4))
plt.plot(full.index, full['aqi_clean'], color='#1FA4B1', linewidth=0.8)
plt.title('AQI Harian Setelah Pembersihan (celah panjang tetap kosong)')
plt.xlabel('Tanggal')
plt.ylabel('AQI')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
#202432096_SHIMA
# Pembentukan sliding window, pembagian data latih/uji, normalisasi, dan konversi ke Tensor
WINDOW = 30  # 30 hari terakhir sebagai input untuk memprediksi hari berikutnya

def create_sequences(values, index, window):
    X, y, d = [], [], []
    for i in range(len(values) - window):
        seg = values[i:i + window + 1]
        if np.isnan(seg).any():      # lewati window yang mengandung celah
            continue
        X.append(seg[:-1])
        y.append(seg[-1])
        d.append(index[i + window])  # tanggal target prediksi
    return np.array(X, dtype=np.float32), np.array(y, dtype=np.float32), np.array(d)

X_raw, y_raw, target_dates = create_sequences(full['aqi_clean'].values, full.index, WINDOW)
print('Total sampel window :', X_raw.shape)

# Pembagian kronologis 70% latih, 10% validasi, 20% uji (tanpa shuffle agar tidak terjadi kebocoran data)
n = len(X_raw)
n_train, n_val = int(n * 0.7), int(n * 0.8)
X_train_raw, X_val_raw, X_test_raw = X_raw[:n_train], X_raw[n_train:n_val], X_raw[n_val:]
y_train_raw, y_val_raw, y_test_raw = y_raw[:n_train], y_raw[n_train:n_val], y_raw[n_val:]
test_dates = target_dates[n_val:]
print('Data latih / validasi / uji :', len(X_train_raw), '/', len(X_val_raw), '/', len(X_test_raw))

# Normalisasi Min-Max; min dan max dihitung HANYA dari data latih
data_min = min(X_train_raw.min(), y_train_raw.min())
data_max = max(X_train_raw.max(), y_train_raw.max())
def scale(a):
    return (a - data_min) / (data_max - data_min)

# Konversi ke PyTorch Tensor dengan bentuk (sampel, window, 1) dan (sampel, 1)
X_train = torch.tensor(scale(X_train_raw)).unsqueeze(-1)
y_train = torch.tensor(scale(y_train_raw)).unsqueeze(-1)
X_val = torch.tensor(scale(X_val_raw)).unsqueeze(-1)
y_val = torch.tensor(scale(y_val_raw)).unsqueeze(-1)
X_test = torch.tensor(scale(X_test_raw)).unsqueeze(-1)
y_test = torch.tensor(scale(y_test_raw)).unsqueeze(-1)
print('X_train:', tuple(X_train.shape), '| y_train:', tuple(y_train.shape))
print('X_val  :', tuple(X_val.shape), '| y_val  :', tuple(y_val.shape))
print('X_test :', tuple(X_test.shape), '| y_test :', tuple(y_test.shape))

batch_size = 32
train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=batch_size, shuffle=True)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432096_SHIMA
class SequenceModel(nn.Module):
    def __init__(self, input_size=1, hidden_size=32, num_layers=2, dropout=0.2):
        super(SequenceModel, self).__init__()
        # Lapisan LSTM bertumpuk; batch_first=True -> input berbentuk (batch, window, fitur)
        self.lstm = nn.LSTM(input_size, hidden_size, num_layers,
                            batch_first=True, dropout=dropout)
        # Lapisan linear untuk mengubah hidden state terakhir menjadi 1 nilai prediksi
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        out, _ = self.lstm(x)               # out: (batch, window, hidden_size)
        delta = self.fc(out[:, -1, :])      # LSTM memprediksi PERUBAHAN AQI dari hari terakhir
        return x[:, -1, :] + delta          # prediksi = nilai hari terakhir + perubahan (koneksi residual)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432096_SHIMA
# 1. Inisiasi hyperparameter
epochs = 100       # batas maksimum; pelatihan dapat berhenti lebih awal (early stopping)
learning_rate = 0.001
hidden_size = 32
num_layers = 2
dropout = 0.2
patience = 10      # berhenti jika validation loss tidak membaik selama 10 epoch

# 2. Inisiasi model, fungsi Loss, dan Optimizer
model = SequenceModel(1, hidden_size, num_layers, dropout).to(device)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)
print(model)

# 3. Simpan riwayat loss dan state model terbaik
train_losses = []
val_losses = []
best_val, best_epoch, best_state, wait = float('inf'), 0, None, 0

# 4. Training loop dengan early stopping
X_val_d, y_val_d = X_val.to(device), y_val.to(device)
for epoch in range(epochs):
    model.train()
    running = 0.0
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()                      # reset gradien
        loss = criterion(model(xb), yb)            # forward pass + hitung loss
        loss.backward()                            # backpropagation
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)  # cegah exploding gradient
        optimizer.step()                           # perbarui bobot
        running += loss.item() * xb.size(0)
    train_losses.append(running / len(train_loader.dataset))

    model.eval()
    with torch.no_grad():
        val_loss = criterion(model(X_val_d), y_val_d).item()
    val_losses.append(val_loss)

    # Simpan bobot terbaik berdasarkan validation loss
    if val_loss < best_val - 1e-6:
        best_val, best_epoch, wait = val_loss, epoch + 1, 0
        best_state = {k: v.detach().clone() for k, v in model.state_dict().items()}
    else:
        wait += 1

    if epoch == 0 or (epoch + 1) % 5 == 0:
        print(f'Epoch {epoch+1:3d}/{epochs} | Train Loss: {train_losses[-1]:.6f} | Val Loss: {val_losses[-1]:.6f}')

    if wait >= patience:
        print(f'Early stopping pada epoch {epoch+1}')
        break

# 5. Pulihkan bobot terbaik
model.load_state_dict(best_state)
print(f'Bobot terbaik dari epoch {best_epoch} dengan Val Loss {best_val:.6f}')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432096_SHIMA
# Prediksi pada data uji, kembalikan ke skala AQI asli, lalu hitung metrik evaluasi
model.eval()
with torch.no_grad():
    pred_scaled = model(X_test.to(device)).cpu().numpy().ravel()
pred = pred_scaled * (data_max - data_min) + data_min
actual = y_test_raw
naive = X_test_raw[:, -1]  # baseline naif: AQI hari kemarin

def evaluate(a, p):
    err = a - p
    mae = np.mean(np.abs(err))
    rmse = np.sqrt(np.mean(err ** 2))
    mape = np.mean(np.abs(err / a)) * 100
    r2 = 1 - np.sum(err ** 2) / np.sum((a - a.mean()) ** 2)
    return {'MAE': mae, 'RMSE': rmse, 'MAPE (%)': mape, 'R2': r2}

results = pd.DataFrame({'LSTM': evaluate(actual, pred),
                        'Baseline Naif (kemarin)': evaluate(actual, naive)}).T
results.round(3)
```

```python
#202432096_SHIMA
# 1. Plot Grafik Penurunan Loss Training terhadap Epoch
plt.figure(figsize=(10, 4))
plt.plot(train_losses, color='#114F6A', linewidth=2, label='Training Loss')
plt.plot(val_losses, color='#1FA4B1', linewidth=2, linestyle='--', label='Validation Loss')
plt.axvline(best_epoch - 1, color='gray', linestyle=':', label=f'Epoch terbaik ({best_epoch})')
plt.title('Grafik Penurunan Loss Selama Pelatihan')
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE, skala ternormalisasi)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

# 2. Plot Grafik Perbandingan Data Aktual vs Hasil Prediksi Model (seluruh data uji)
plt.figure(figsize=(12, 5))
plt.plot(test_dates, actual, color='#114F6A', linewidth=1.5, label='Data Aktual')
plt.plot(test_dates, pred, color='#1FA4B1', linewidth=1.5, label='Prediksi LSTM')
plt.title('Data Aktual vs Prediksi LSTM pada Data Uji')
plt.xlabel('Tanggal')
plt.ylabel('AQI')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

# 3. Zoom pada 90 data uji terakhir agar selisih lebih jelas
n = 90
plt.figure(figsize=(12, 5))
plt.plot(test_dates[-n:], actual[-n:], color='#114F6A', marker='o', markersize=3, label='Data Aktual')
plt.plot(test_dates[-n:], pred[-n:], color='#1FA4B1', marker='o', markersize=3, label='Prediksi LSTM')
plt.title(f'Zoom: {n} Data Uji Terakhir')
plt.xlabel('Tanggal')
plt.ylabel('AQI')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

Training loss turun perlahan dari sekitar 0,0029 menjadi sekitar 0,0025, sedangkan validation loss berada di kisaran 0,0007 dan hampir datar. Early stopping aktif pada epoch 25 dan bobot terbaik diambil dari epoch 15 (validation loss 0,000680), sehingga model tidak dibiarkan overfitting. Training loss lebih tinggi daripada validation loss, kemungkinan karena data latih memuat periode dengan fluktuasi AQI yang lebih besar dan dropout aktif saat pelatihan. Pada data uji, LSTM mencapai MAE 5,143, RMSE 8,257, MAPE 8,472%, dan R² 0,819, sedikit lebih baik daripada baseline naif (MAE 5,291; RMSE 8,328; MAPE 8,748%; R² 0,815). Selisihnya kecil, sehingga model pada dasarnya bertumpu pada nilai hari sebelumnya dengan koreksi kecil. Grafik menunjukkan prediksi mengikuti pola data aktual, termasuk lonjakan besar sekitar September 2019 (AQI ±187). Pada grafik zoom, prediksi cenderung tertinggal sekitar satu hari dan kurang menangkap kenaikan atau penurunan tajam yang mendadak. Garis lurus pada grafik sekitar Desember 2019 sampai Februari 2020 muncul karena data periode itu dihapus (nilai konstan 52), sehingga tidak ada window valid di sana. Itu bukan hasil prediksi model.
